```python
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='Gender', palette='Set2')
plt.title('Distribution of Gender')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(data=df, x='Age', bins=10, kde=True, color='skyblue')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='ResidenceCountry', palette='pastel')
plt.title('Users by Country')
plt.xlabel('Country')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.countplot(data=df, x='KycStatusLabel', palette='coolwarm', order=df['KycStatusLabel'].value_counts().index)
plt.title('KYC Status Breakdown')
plt.xlabel('KYC Status')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='CompletedProfile', palette='Set1')
plt.title('Completed vs Incomplete Profiles')
plt.xlabel('Completed Profile')
plt.ylabel('Count')
plt.show()
```

```python
occupation_industry = pd.crosstab(df['Occupation'], df['Industry'])
plt.figure(figsize=(12,8))
sns.heatmap(occupation_industry, cmap='Blues', annot=True, fmt='d')
plt.title('Occupation vs Industry')
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df, x='Generation', palette='viridis', order=df['Generation'].value_counts().index)
plt.title('Generation Distribution')
plt.xlabel('Generation')
plt.ylabel('Count')
plt.show()
```
