```python
import os

url = "https://www.propertypro.ng/"

response = requests.get(url)

if response.status_code == 200:
    soup = BeautifulSoup(response.content, 'html.parser')

    # Scrape images
    image_folder = 'images'
    os.makedirs(image_folder, exist_ok=True)
    image_tags = soup.find_all('img')

    for img_tag in image_tags:
        img_url = img_tag['src']
        if not img_url.startswith('http'):
            img_url = url + img_url
        img_response = requests.get(img_url)

        # Save the image to the images folder
        with open(os.path.join(image_folder, os.path.basename(img_url)), 'wb') as img_file:
            img_file.write(img_response.content)

    # Scrape text
    text_data = []
    text_tags = soup.find_all(['p', 'h1', 'h2', 'h3', 'h4', 'h5', 'h6', 'span', 'div'])

    for tag in text_tags:
        text_data.append(tag.get_text())

    # Print the text data
    print("\n".join(text_data))

else:
    print(f"Failed to fetch data from {url}. Status code: {response.status_code}")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import numpy as np

# Suppose cleaned_text_str is your scraped data as one long string
documents = [cleaned_text_str]  # Treat entire text as a single document

vectorizer = TfidfVectorizer(stop_words="english")
tfidf_matrix = vectorizer.fit_transform(documents)

# Extract vocabulary and scores
feature_names = vectorizer.get_feature_names_out()
tfidf_scores = tfidf_matrix.toarray()[0]
```

```python
def word_similarity(word1, word2):
    # Ensure lowercase matching
    word1, word2 = word1.lower(), word2.lower()

    if word1 not in feature_names or word2 not in feature_names:
        return None  # one of the words isn't in the text

    # Find their TF-IDF scores
    v1 = tfidf_scores[np.where(feature_names == word1)[0][0]]
    v2 = tfidf_scores[np.where(feature_names == word2)[0][0]]

    # Similarity = 1 - normalized absolute difference
    max_score = max(v1, v2) if max(v1, v2) > 0 else 1
    similarity = 1 - abs(v1 - v2) / max_score
    return round(similarity, 3)
```

```python
print("Similarity between 'book' and 'novel':", word_similarity("book", "novel"))
print("Similarity between 'book' and 'price':", word_similarity("book", "price"))
print("Similarity between 'fiction' and 'fantasy':", word_similarity("fiction", "fantasy"))

```

```python
import requests
from bs4 import BeautifulSoup
import re

url = "http://books.toscrape.com/"
response = requests.get(url)

if response.status_code == 200:
    soup = BeautifulSoup(response.text, "html.parser")
    books = soup.find_all("article", class_="product_pod")
    scraped_text = ""

    for book in books:
        title = book.h3.a["title"]
        price = book.find("p", class_="price_color").text
        scraped_text += f"{title} {price}\n"

    # Clean extra whitespace
    cleaned_text_str = re.sub(r"\s+", " ", scraped_text).strip()
else:
    print("Failed to retrieve the page.")
```

```python
import nltk
from nltk.tokenize import word_tokenize
nltk.download('punkt')

sentences = [word_tokenize(line.lower())
             for line in cleaned_text_str.split("\n") if line.strip()]
```

```python
from gensim.models import Word2Vec

model = Word2Vec(
    sentences,
    vector_size=50,  # embedding size
    window=3,        # context window
    min_count=1,     # include all words
    workers=4,
    sg=1             # skip-gram for better quality on small data
)

model.save("books_word2vec.model")
print("Model trained and saved.")
```

```python
# Get vector for a word
print(model.wv["book"])  # if "book" exists

# Find similar words
print(model.wv.most_similar("book", topn=5))
```
